#what is Pandas?

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, built on the top of the python programming language.

**Pandas Series**

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type. ex column like :name

#Importing pandas

```python
import numpy as np
import pandas as pd
```

#Creating Series from lists.

```python
#string
country = ['India', 'USA', 'London', 'Nepal']
pd.Series(country)

#Series contains : index(automatically generates) and value(we provide)
```

Integers

```python
#integers
runs = [13, 14, 65, 100, 78]
runs_series = pd.Series(runs)
```

```python
#Custom index : we can make our own custom index
#ex: subject name as index

marks = [67, 8, 9, 50]
subjects = ['eng', 'maths', 'sci', 'geog']

pd.Series(marks, index=subjects)
```

```python
#Setting a name to your series
marks = pd.Series(marks, index=subjects, name='Bhargavi ke marks')
marks
```

Series from dict

```python
marks = {'eng':67, 'maths':8, 'sci':9, 'geog':50}
marks_series = pd.Series(marks, name= 'Bhargavi ke marks')
marks_series
```
