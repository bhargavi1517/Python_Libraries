```python
#find all the final winners
#final matches are labelled
mask = ipl['MatchNumber'] == 'Final'
new_df = ipl[mask]
new_df[['Season', 'WinningTeam']]
```

```python
ipl[ipl['MatchNumber'] == 'Final'][['Season', 'WinningTeam']].shape[0]
```

```python
# how many super over finishes have occured :-> same runs
ipl
ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
#how many matches has csk won in kolkatta
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]
#Bitwise and when we work with binary
```

```python
#toss winner is match winner in percentage
(ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0])*100
```

```python
#movies with ratings higher than 8 and votes> 10000
movies.head()
movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)].shape[0]
```

```python
#Action movies with rating higher than 7.5
mask1=movies['genres'].str.split('|').apply(lambda x: 'Action' in x)
movies[mask1 & (movies['imdb_rating'] > 7.5)].shape[0]
```
